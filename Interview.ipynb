```python
#palindrome

string = "MalayalaM"
revstr = string[::-1]
print(string)
print(revstr)

if(string == revstr):
    print("True")
else:
    print("False")
```

```python
mystr = "Malay alam"
newstr = mystr.lower().replace(' ',' ')

revstr =newstr[::-1]
print(revstr)

if newstr == revstr:
    print("True")
else:
    print("False")
    
```

```python
mystr = "Hello worlD"
newstr = ''
for each in mystr:
    if each.isalpha():
        newstr = newstr + each.lower()
        
mydict = {}

for each_1 in newstr:
    if each_1 not in mydict.keys():
        mydict[each_1] = 1
    else:
        mydict[each_1]+=1
print(mydict)

        
```

```python
#task m
#output =[100,"hello",20.55,True,False,"Hello","Hai"]
def remove_duplicate(mylist):
        original = []

        for each in mylist:
             if each not in original:
                  original += [each]
        return original
mylist = [100,"hello",20.55,"hello",100,True,False,100,"Hello","Hai",True,"hello"]
print(remove_duplicate(mylist))
        
```
