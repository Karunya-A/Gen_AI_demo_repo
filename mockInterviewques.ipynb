```python
s = input("Enter a string:")

upper = ""
lower = ""
digits = ""
specialcharacters = ""

for each in s:
    if each.isupper():
        upper = upper + each
    elif each.islower():
        lower = lower + each
    elif each.isdigit():
        digits =  digits + each
    else:
        specialcharacters = specialcharacters + each
        
print("Upper:" ,upper)
print("lower:" ,lower)
print("digits:" ,digits)
print("specialcharacters:" ,specialcharacters)
```
