```python
n1 = int(input("Enter a number: "))
print("Enter a number", n1)

reverse = 0

while n1 > 0:
    remainder = n1 % 10
    reverse = reverse * 10 + remainder
    n1 = n1 // 10

print("Reversed number is:", reverse)
```

```python
n = int(input("Enter a number"))
print("Enter a number:",n)

fact = 1

while n > 0:
    fact = fact * n
    n= n-1
    
    
print("Factorial is:", fact)
```
