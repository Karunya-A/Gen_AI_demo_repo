```python
def strupper(func):
    def inner():
        str1 = func()
        return str1.upper()
    return inner

def strtitle(func):
    def inner():
        str1 = func()
        return str1.title()
    return inner

@strtitle
def strname():
    return "kArunya"

@strupper
def myname():
    return "Kani ManI"


@strupper
def yourname():
    return "Jeni BERCY"

print(strname())
print(myname())
print(yourname())

```

```python
#decorator with parameter

def divdec(func):
    def inner(x,y):
        if y == 0:
            return "Don't get divisior as zero"
        else:
            return func(x,y)
    return inner

@divdec
def div(a,b):
    return a/b


print(div(8,2))
print(div(4,0))

```

```python
#instance method

class subject:
    def __init__(self,Tamil,English,Maths):
        self.tamil = Tamil
        self.english = English
        self.maths = Maths
        
    def marks(self):
        Total =  self.tamil + self.english +self.maths
        return Total


Kavi = subject(10,40,80)
Hema = subject(50,90,78)
subha = subject(70,54,32)

result1 = Kavi.marks()
result2 = Hema.marks()
print(result1)
print(result2)
subha.marks()
```
