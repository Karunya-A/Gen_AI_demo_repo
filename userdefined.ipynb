```python
def add_numbers(a, b):
    result = a + b
    return result

num1 = int(input("Enter first number: "))
num2 = int(input("Enter second number: "))

answer = add_numbers(num1, num2)

print("Sum =", answer)
```

```python
def myfunc(x,y):
    print("X =", x)
    print("Y=", y)
    print("The sum of x + y is =" ,x+y)

myfunc(10,20) 
print()
```

```python
#function with parameters

def myfunc(x = 0, y = 0):
    print("X =", x)
    print("Y =" ,y)
    print("The sum is:" ,x+y)
    
myfunc()
print()
myfunc(20,30)
print()
myfunc(750)
print()
```

```python
#function with return keyword

def myfunc(x = 0, y = 0):
    print("X =", x)
    print("Y =" ,y)
    print("The sum is:" ,x+y)

ans = myfunc(100,200)
print(ans)
```

```python
def myadd(x= 0,y =0):
    return x+y
def mysub(x = 0, y = 0):
    return x-y
ans = myadd(50,70) - mysub(40,10)
print(ans)
```

```python
def myadd(x= 0,y =0):
    return x+y
def mysub(x = 0, y = 0):
    return x-y

ans = mysub(myadd(60,30), mysub(30,10))
print(ans)
```

```python
#Arbitrary arguments *args - arguments stored in tuples

def myfunc(*x):
    print(x , type(x))

myfunc(10)
myfunc("hello")
myfunc(True)
myfunc(60.8)
print("----------------")
myfunc(10,"Hi",4.6)
```

```python
#**kwargs - arguments stored in dictinory
def myfunc(**x):  
    print(x, type(x))
myfunc(a = 10)
myfunc(name = "Karunya" ,age = 22)
```
