```python
#keys

mydict = {
    "Name" :"Karunya",
    "City" :"London",
    "Tech" :"Python with GEN AI",
    900 :800
    
}
print(mydict)
print(mydict.keys())
print(list(mydict.keys()))
print()

for each in mydict.keys():
    print(each)
```
