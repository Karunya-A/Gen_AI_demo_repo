```python
a = int(input("Enter a number:"))
symbol = input("Enter a symbol:")

for i in range(a,0,-1):
    for j in range(i):
        print(symbol,end=' ')
    print()
```

```python
n = int(input("Enter a number of rows:"))

for i in range(n):
    for j in range(1,i+1):
        print(i,end=' ')
    print()
        
        
        
```

```python
n = int(input("Enter a rows:"))
symbol = input("Enter a columns:")

for i in range(1,n+1):
        print(' '*(n-i) + symbol *i)
    
```
