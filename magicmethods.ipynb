```python
class Myclass:
    def __init__(self):
        print("Constructor calling")
    def __del__(self):
        print("Destructor calling")
    def __str__(self):
        return "<class 'Myclass'>"
        
obj = Myclass()
print(obj)
```

```python
def myfunc():
    print("Constructor calling")
def newfunc():
    print("Hello it's my function")

if __name__ == "__main__" :
    myfunc()
    newfunc()
```

```python
# Operator overloading with magic methods

class Myclass:  
    
    def __init__(self,x):
        self.x = x
    def __add__(self,others):
        return(f"The total of {self.x} and {others.x} is ="  +  str (self.x +others.x))
        

obj1 = Myclass(30)
obj2 = Myclass(40)

print(obj1 + obj2)
    
```

```python
class Myclass:  
    
    def __init__(self,x):
        self.x = x
    def __truediv__(self,others):
        return(f"The division of {self.x} and {others.x} is ="  +  str (self.x /others.x))
        

obj1 = Myclass(30)
obj2 = Myclass(40)

print(obj1 / obj2)
    
```
