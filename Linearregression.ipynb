```python
input_data = car_data.drop(columns = ["year" ,"price_category"])
output_data = car_data["price_category"]
input_data
```

```python
output_data
```

```python
print(input_data.shape)
print(output_data.shape)
```

```python
import pandas as pd
input_data = pd.get_dummies(input_data ,dtype = int)
print(input_data.head())
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
input_data  =scaler.fit_transform(input_data)
print(input_data)
```

```python
from sklearn.model_selection import train_test_split

input_data_train, input_data_test, output_data_train, output_data_test = train_test_split(
    input_data,
    output_data,
    test_size=0.2,
    random_state=42
)
```

```python
print(input_data_train.shape)
print(input_data_test.shape)
```

```python
from sklearn.linear_model import LinearRegression
```

```python
from sklearn import metrics
model = LinearRegression()
model.fit(input_data_train ,output_data_train)
```

```python
predicted_km_driven = model.predict(input_data_test)
predicted_km_driven
```

```python
from sklearn.metrics import mean_absolute_error ,mean_squared_error
```

```python
print("mean absolute error:" ,(metrics.mean_absolute_error(output_data_test ,predicted_km_driven)))
```

```python
print("mean squared error:" ,(metrics.mean_squared_error(output_data_test,predicted_km_driven)))
```
