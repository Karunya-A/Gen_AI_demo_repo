```python
while True:
    try:
        num = int(input("Enter a number in between 20 to 100:"))
        if num<20:
            raise ZeroDivisionError
        elif num>100:
            raise ValueError
        else:
            print(f"The entered number is:{num}")
            break

    except ValueError:
        print("Enter the number lesser than 100 ")
    except ZeroDivisionError:
        print("Enter the number greater than 20")
    
print("Program done")

    
```

```python
try:
    num = int(input("Enter a number btwn 30 to 45:"))
    if num <30:
        raise ValueError("Enter the number greater than 30  ")
    elif num >45:
        raise ZeroDivisionError("Enter the number lesser than 45 ")
    else:
        print(f"The entered number is:{num}")
    
except ValueError as x:
    print(x)
except ZeroDivisionError as y:
    print(y)
    
    
print("----------")
```

```python

try:
    print("Outer block")
    
    try:
        num = int(input("Enter a number:"))
        result = 10/num
        print(result)
        
    except ValueError:
        print("Inner error: Enter a correct input")
    except ZeroDivisionError:
        print("Outer error: Dont divide  by zero")
except Exception as e:
    print("Outer error",e)
        
```
