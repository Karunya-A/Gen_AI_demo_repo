```python
import json
dict_data = {
    "name": "Karunya",
    "age": 30,
    "city": ("New York","paris"),
    "year": None
}

json_string = json.dumps(dict_data)
print(json_string, type(json_string))

dict_from_json = json.loads(json_string)
print(dict_from_json, type(dict_from_json))
```

```python
import json

student = {
    "Name" : "Karunya",
    "Course Enrolled": "Python GenAI",
    "Age" : 22,
    "District" : "Kerala"
}

with open("student.json" ,"w") as file:
    json.dump(student,file)

print("Data saved successfully")
```

```python
import json

student = {
    "Name" : "Karunya",
    "Course Enrolled": "Python GenAI",
    "Age" : 22,
    "District" : "Kerala"
}

with open("student.json" ,"r") as file:
    student = json.load(file)

print(student)
print(student["Age"])
```
