```python
#OOPS - Polymorphism

class fruit:
    def mango(self):
        print("Its yellow  in color")
    def strawberry(self):
        print("Its red in color")
    def grapes(self):
        print("Its green in color")

class vegetables:
    def mango(self):  
        print("Its both vegetables and fruits")    
    def strawberry(self):
        print("It  is only fruit")
    def grapes(self):
        print("Its not vegetables category")
        
obj_fruit = fruit()
obj_vegetables = vegetables()
    
for food in [obj_fruit , obj_vegetables]:
    food.mango()
    food.strawberry()
    food.grapes()

    print()
```

```python
#Encapsulation - private variable cannot be changed

class variable:
    a = 100
    b= 200
    
obj = variable()
print(obj.a , obj.b)
obj.b = 300
print(obj.a , obj.b)
```
