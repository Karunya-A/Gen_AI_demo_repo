```python
import os 
import csv
ip = []
http_status = []
mydict = {}

path ="C:\\Users\\DELL\\Downloads\\Python_Learning"
filename = "mylog.log"

with open(os.path.join(path,filename)) as f:
    for each_line in f:
        mylist = each_line . split(" ")
        ip.append(mylist[3]) 
        http_status.append(mylist[8])

 

def frequency(mylist):
    
    for each in mylist:
        if each not in mydict:
            mydict[each] = 1
        else:
            mydict[each] +=1
    return mydict


def write_to_csv(mydict):
    with open(os.path.join(path,"new_csv.csv"),"w",newline = "")as f:
        writer = csv.writer(f)
        writer.writerow(["Sno","http_status","Frequncy"])
        Sno = 1
        for each in mydict:
            writer.writerow([Sno,each,mydict[each]])
            Sno +=1
  
rest1 = frequency(ip)
rest2 = frequency(http_status )

write_to_csv(rest1)
write_to_csv(rest2)

print("CSV file created successfully")
```

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("new_csv.csv")

data = df[df["http_status"].isin(["200", "304", "404","10.73.144.159:53104"])]

plt.bar(data["http_status"], data["Frequncy"],
        color=["skyblue", "orange", "lightgreen","blue"], edgecolor="black",)

plt.xlabel("HTTP Status Code")
plt.ylabel("Frequency")
plt.title("HTTP Bar Graph")

plt.show()
```
