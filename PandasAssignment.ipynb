```python
def salary_level(salary):
    if salary >= 50000:                   #8
        return "High"
    elif salary > 40000:
        return "Medium"
    else:
        return "Low"

df["SalaryLevel"] = df["Salary"].apply(salary_level)

print(df[["Name" ,"Salary" ,"SalaryLevel"]])  
```

```python
marks = pd.DataFrame({
    "EmpID" :[101,103,105,108,110],
    "TrainingScore" :[89,54,76,96,63]
})
                                                                #9
result = df.merge(marks,on = "EmpID" ,how = "left")

print(result[["EmpID" ,"Name" ,"TrainingScore"]])
```
