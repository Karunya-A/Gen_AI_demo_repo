```python
##Scatter plot

import numpy as np

x = np.array([1,2,3,4,5,1])
y = np.array([1,2,3,2,4,6])

plt.scatter(x,y,color = ["r","g","r","y","g","r"],marker = "^",s = 150)
plt.show();
```

```python
# Line plot

import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 4, 5, 1])
y = np.array([1, 2, 3, 2, 4, 6])

c = ['red', 'green', 'yellow', 'pink', 'blue', 'white']

plt.plot(x ,y ,c = 'g',marker = "*" ,linestyle = "dashdot")
plt.scatter(x, y, color=c, marker="*")
plt.show()
```

```python
import matplotlib.pyplot as plt

year = ["1990", "2010", "2020", "2007"]
markerpoints = [10, 20, 30, 40]
c= ["pink" ,"blue" ,"yellow" ,"orange"]

plt.bar(year, markerpoints,color = c,align = "edge",edgecolor = "green")
plt.show();
```

```python
# Frequency Distribution

np.random.seed(4)

a = np.random.normal(50,1,100)


plt.hist(a,color = "blue",bins = 15,edgecolor = "black")
plt.show();
```

```python
# Pie Chart 

numbers  = [10,20,45,78]
points = [100,20,23,76]

exp = [0.05,0,0,0]
c= ["pink" ,"blue" ,"yellow" ,"orange"]

plt.pie(points ,labels = numbers , colors = c,explode = exp)
plt.show();
```

```python
#Stock Market

stockA = [1,2,4,5]
stockB = [3,6,8,2]
stockC = [5,9,2,1]

plt.title("Stock Market")

plt.plot(stockA,label  ="StockA")
plt.plot(stockB ,label = "StockB")
plt.plot(stockC,label = "StockC")
plt.legend()
plt.show();
```

```python
# Method (pyplot API)

fig = plt.figure()
ax = fig.add_subplot()

ax.plot([1,2,3,4,5] ,[-2,-1,3,7,8],c = "r")
plt.show();
```

```python
# object oriented API

fig,ax = plt.subplots()
ax.plot([1,2,3] ,[5,6,7])
plt.show()
```

```python
# pit from dict

car_mileage = {
    "BMW" :7,
    "Toyota" :10,
    "Porsch" :3
}

fig,ax = plt.subplots()

ax.bar(car_mileage.keys() ,car_mileage.values())
plt.show();
```
