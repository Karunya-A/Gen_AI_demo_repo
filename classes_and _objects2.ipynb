```python
#Single Inheritance

class A:
    a = 100
    
    def myfunc(self):
        print("Hello my function is calling")
        print(f"The variable for myfunction is:{self.a}")
        
class B(A):
    b =200
    
    def newfunc(self):
            print("Hello my function is calling")
            print(f"The variable for myfunction is:{self.b}")

obj = B()
obj.myfunc()
print()
obj.newfunc()
    
    
```

```python
# Multiple Inheritance 

class A:
    a = 100
    
    def myfunc(self):
        print("Hello my function is calling")
        print(f"The variable for myfunction is:{self.a}")
        
class B:
    b =200
    
    def newfunc(self):
            print("Hello my function is calling")
            print(f"The variable for myfunction is:{self.b}")
class C(B,A):
    c = 300
    
    def oldfunc(self):
        print("Hello my function is calling")
        print(f"The variable for myfunction is:{self.c}")

obj = C()
obj.myfunc()
print()
obj.newfunc()
print()
obj.oldfunc()

        

```

```python
#Super function

class A:
    a = 100
    
    def myfunc(self):
        print("Hello my function is calling")
        print(f"The variable for myfunction is:{self.a}")
        
        
class B(A):
    b =200
    
    def myfunc(self):
            print("Hello my function is calling")
            print(f"The variable for myfunction is:{self.b}")
            super().myfunc()
class C(B):
    c = 300
    
    def myfunc(self):
        print("Hello my function is calling")
        print(f"The variable for myfunction is:{self.c}")
        super().myfunc()
        
obj = C()
obj.myfunc()
```
