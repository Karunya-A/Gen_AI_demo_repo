```python
import pandas as pd

data = {
    "Name": ["Kavi", "Suvi", "Puvi", "Ravi", "Kavi"],
    "Age": [23, None, 12, 25, 23],
    "Salary": [25000, 35000, None, 30000, 25000]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.isnull())
print()
print(df.isnull().sum())
print()
print(df.dropna())
print()
print(df.fillna(0))
print()
print(df.duplicated())
print()
```

```python
import pandas as pd

data = {
    "Name": ["Kavi", "Suvi", "Puvi", "Ravi", "Abi"],
    "Department": ["IT", "HR", "IT", "HR", "IT"],
    "Salary": [25000, 35000, 15000, 30000, 40000]
}

df = pd.DataFrame(data)
print(df)
```

```python
print(df.sort_values("Salary"))
print()
print(df.sort_values("Salary",ascending = False))
print()
print(df.groupby("Department")["Salary"].sum())
print()
print(df.groupby("Department").sum())
print()
print(df.value_counts("Department"))
print()
print(df["Department"].unique())
print()
print(df["Department"].nunique())
```
